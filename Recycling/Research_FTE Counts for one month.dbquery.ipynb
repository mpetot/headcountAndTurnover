```sql

SELECT
  f.dateint,
  f.LegalEntityShort,
  f.jobcodecategory,
  f.CostType,
  f.EPMBranch,
  f.Breakdown2Id,
  f.BreakDownTwo,
  f.IsOverScale,
  b.alias,
  sum(f.FTE) as sumfte
FROM
  `na-cdp-bu-hr-dev`.curated.fte as f
    LEFT JOIN (
      SELECT
        backendname,
        alias
      FROM
        `na-bu-cdm-dev`.staging_usdwh_tkefinance.epmdimensionmetadata
    ) as b on b.backendname = f.EPMBranch
WHERE
  DateInt >= 20251200 and DateInt <20260000 and fte <> 0
  and LegalEntityShort = 'Canada FLD'
GROUP BY
  ALL;

```
