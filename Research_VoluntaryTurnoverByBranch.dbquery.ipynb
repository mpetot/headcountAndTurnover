```sql
SELECT
  TRUNC(TO_DATE(CAST(dateint AS STRING), 'yyyyMMdd'), 'MONTH') as `Calendar Date`,
  Country,
  EPMBranch,
  branch.alias as branchname,
  CostType,
  FTEVoluntaryTurnover
FROM
  `NA-CDP-BU-HR-DEV`.curated.voluntaryturnover as vol
    LEFT JOIN (
      SELECT
        backendname,
        alias
      FROM
        `na-bu-cdm-dev`.staging_usdwh_tkefinance.epmdimensionmetadata WHERE Source = 'EPMEntity'
    ) as branch on branch.backendname = vol.epmbranch
```
